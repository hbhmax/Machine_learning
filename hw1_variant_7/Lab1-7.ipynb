1. What is the result of the following expression?

```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
print('0 * np.nan            =', 0 * np.nan)
print('np.nan == np.nan     =', np.nan == np.nan)
print('np.inf > np.nan      =', np.inf > np.nan)
print('np.nan - np.nan      =', np.nan - np.nan)
print('np.nan in set([np.nan]) =', np.nan in set([np.nan]))
print('0.3 == 3 * 0.1       =', 0.3 == 3 * 0.1)
print('3 * 0.1              =', repr(3 * 0.1))
````

**Ответ:** `nan`, `False`, `False`, `nan`, `True`, `False`.

- Любая арифметика с `nan` даёт `nan`, а любое сравнение с `nan` (даже `nan == nan`) даёт `False`.
- `np.nan in set([np.nan])` даёт `True`, потому что `in` для множеств сначала проверяет *идентичность* объектов (`is`), и только потом равенство. Здесь это один и тот же объект `np.nan`.
- `0.3 == 3 * 0.1` даёт `False` из-за погрешности представления float: `3 * 0.1 = 0.30000000000000004`. Для сравнения чисел с плавающей точкой используют `np.isclose`.

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
high = df[df['Stress Levels'] >= 7]
low = df[df['Stress Levels'] <= 3]
bp_cols = ['Systolic', 'Diastolic']

print(f'Высокий стресс: n={len(high)}, низкий стресс: n={len(low)}\n')
print(pd.DataFrame({'Высокий стресс': high[bp_cols].mean(),
                    'Низкий стресс': low[bp_cols].mean()}))

print()
for col in bp_cols:
    t = stats.ttest_ind(high[col], low[col], equal_var=False, alternative='greater')
    print(f'{col}: p-value = {t.pvalue:.3f}')

print('\nКорреляция стресса с давлением:')
print(df[['Stress Levels', 'Systolic', 'Diastolic']].corr().iloc[0, 1:].round(3))
````

**Ответ:** нет, неправда. Средние значения почти совпадают (систолическое ≈ 145.9 против 145.5 мм рт. ст., диастолическое ≈ 95.7 против 95.2), разница статистически незначима (p ≈ 0.31 и 0.18), корреляция стресса с давлением ≈ 0.02–0.03, то есть связи практически нет.

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
women = df[df['Gender'] == 'Female']
young = women.loc[women['Age (years)'] <= 30, 'Cognitive Function']
old = women.loc[women['Age (years)'] >= 60, 'Cognitive Function']

print(f'Молодые (<=30): n={len(young)}, средняя самооценка = {young.mean():.2f}')
print(f'Пожилые (>=60): n={len(old)}, средняя самооценка = {old.mean():.2f}')

t = stats.ttest_ind(old, young, equal_var=False, alternative='less')
print(f'\nt-тест Уэлча (односторонний): p-value = {t.pvalue:.2e}')
````

**Ответ:** да, правда. Женщины 60+ оценивают свои когнитивные способности заметно ниже (≈ 57.7), чем женщины до 30 лет (≈ 72.5). Разница около 15 баллов, p-value ≈ 1e-64.

6. Влияет ли физическая активность на вес в возрасте более 50 лет?

````python
over50 = df[df['Age (years)'] > 50]
order = ['Low', 'Moderate', 'High']
groups = [over50.loc[over50['Physical Activity Level'] == lvl, 'Weight (kg)'] for lvl in order]

print(over50.groupby('Physical Activity Level')['Weight (kg)']
            .agg(['count', 'mean', 'median']).loc[order])

print(f'\nANOVA:          p-value = {stats.f_oneway(*groups).pvalue:.2e}')
print(f'Краскел-Уоллис: p-value = {stats.kruskal(*groups).pvalue:.2e}')
````

**Ответ:** да, влияет. В возрасте старше 50 лет средний вес снижается с ростом активности: Low ≈ 74.4 кг, Moderate ≈ 72.1 кг, High ≈ 70.4 кг. Различия значимы (ANOVA p ≈ 4e-5, критерий Краскела–Уоллиса p ≈ 8e-6). Это связь, а не доказательство причинности.

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
q30, q60, q75 = df['BMI'].quantile([0.30, 0.60, 0.75])
print(f'Перцентили BMI: 30% = {q30:.2f}, 60% = {q60:.2f}, 75% = {q75:.2f}')

df['has_chronic'] = df['Chronic Diseases'] != 'None'

high_bmi = df[df['BMI'] > q75]
mid_bmi = df[(df['BMI'] >= q30) & (df['BMI'] <= q60)]

print(f'\nBMI > 75-го перцентиля:      n={len(high_bmi)}, доля с хр. заболеваниями = {high_bmi.has_chronic.mean():.3f}')
print(f'BMI между 30 и 60 перцентилем: n={len(mid_bmi)}, доля с хр. заболеваниями = {mid_bmi.has_chronic.mean():.3f}')

table = pd.crosstab(np.r_[['high'] * len(high_bmi), ['mid'] * len(mid_bmi)],
                    np.r_[high_bmi.has_chronic, mid_bmi.has_chronic])
print(f'\nХи-квадрат: p-value = {stats.chi2_contingency(table)[1]:.2e}')
````

**Ответ:** да, правда. Среди людей с BMI выше 75-го перцентиля хронические заболевания есть у ≈ 70.5%, а среди людей с BMI между 30-м и 60-м перцентилем — у ≈ 54.0%. Разница значима (p ≈ 9e-12).

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
d = df[df['Income Level'].isin(['High', 'Low'])].copy()
d['pa_score'] = d['Physical Activity Level'].map({'Low': 0, 'Moderate': 1, 'High': 2})

print(pd.crosstab(d['Income Level'], d['Physical Activity Level'], normalize='index')
        [['Low', 'Moderate', 'High']].round(3))
print()
print(d.groupby('Income Level')['pa_score'].mean())

hi_inc = d.loc[d['Income Level'] == 'High', 'pa_score']
lo_inc = d.loc[d['Income Level'] == 'Low', 'pa_score']
u = stats.mannwhitneyu(hi_inc, lo_inc, alternative='greater')
chi = stats.chi2_contingency(pd.crosstab(d['Income Level'], d['Physical Activity Level']))
print(f'\nМанн–Уитни (H1: у High-дохода активность выше): p-value = {u.pvalue:.3f}')
print(f'Хи-квадрат на независимость:                    p-value = {chi[1]:.3f}')
````

**Ответ:** нет, неправда. Распределения активности у людей с высоким и низким доходом почти одинаковы (высокая активность: ≈ 21% против ≈ 23%; средний балл активности даже чуть ниже у высокого дохода: 0.89 против 0.93). Статистически различий нет (p ≈ 0.90 для односторонней гипотезы, p ≈ 0.45 для хи-квадрата).

6. Имеет ли связь зрение и способность слышать звуки на высоких частотах? Ответьте на вопрос - какая зависимость между ними и предположите почему зависимость именно такая?

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

sc = axes[0].scatter(df['Vision Sharpness'], df['Hearing Ability (dB)'],
                     c=df['Age (years)'], cmap='viridis', alpha=0.6, s=14)
plt.colorbar(sc, ax=axes[0], label='Возраст')
r = df['Vision Sharpness'].corr(df['Hearing Ability (dB)'])
axes[0].set_title(f'Зрение и слух (r = {r:.2f})')
axes[0].set_xlabel('Vision Sharpness (больше = лучше)')
axes[0].set_ylabel('Hearing Ability, dB (больше = хуже порог слуха)')

def residuals(y, x):
    k, b = np.polyfit(x, y, 1)
    return y - (k * x + b)

rv = residuals(df['Vision Sharpness'], df['Age (years)'])
rh = residuals(df['Hearing Ability (dB)'], df['Age (years)'])
axes[1].scatter(rv, rh, alpha=0.4, s=14, color='slategray')
axes[1].set_title(f'После учёта возраста (частная корреляция = {np.corrcoef(rv, rh)[0, 1]:.2f})')
axes[1].set_xlabel('Остатки зрения')
axes[1].set_ylabel('Остатки слуха')
plt.tight_layout()
plt.show()

print(df[['Vision Sharpness', 'Hearing Ability (dB)', 'Age (years)']].corr().round(2))
````

**Ответ:** да, связь есть: **отрицательная, достаточно сильная** (r ≈ −0.64). Чем лучше зрение, тем ниже порог слышимости в dB (то есть тем лучше слух), и наоборот. Заметно также, что у многих людей зрение упёрлось в минимум 0.2 (вертикальная полоса слева), такие значения нужно считать «обрезанными».

**Почему так.** Вероятнее всего, это *не* прямая зависимость, а влияние общей причины, **возраста**. С возрастом зрение падает (r ≈ −0.90), а порог слышимости растёт, то есть слух ухудшается (r ≈ +0.71). Поэтому у пожилых людей одновременно плохое зрение и плохой слух. Если убрать влияние возраста (правый график), частная корреляция становится ≈ 0.01: при одном и том же возрасте зрение и слух друг с другом не связаны. Это пример ложной (confounded) корреляции.

7. Красиво визуализируйте как связаны между собой уровень дохода, когнитивные способности и уровень стресса*

````python
fig, axes = plt.subplots(1, 2, figsize=(15, 6), gridspec_kw={'width_ratios': [1.3, 1]})
palette = {'Low': '#e76f51', 'Medium': '#e9c46a', 'High': '#2a9d8f'}
order = ['Low', 'Medium', 'High']

sns.scatterplot(data=df, x='Stress Levels', y='Cognitive Function', hue='Income Level',
                hue_order=order, palette=palette, alpha=0.3, s=18, ax=axes[0], legend=False)
for lvl in order:
    sub = df[df['Income Level'] == lvl]
    k, b = np.polyfit(sub['Stress Levels'], sub['Cognitive Function'], 1)
    xs = np.linspace(1, 10, 50)
    axes[0].plot(xs, k * xs + b, color=palette[lvl], lw=3, label=f'{lvl} (наклон {k:+.2f})')
axes[0].legend(title='Доход')
axes[0].set_title('Когнитивные способности vs стресс по группам дохода')
axes[0].set_xlabel('Уровень стресса')
axes[0].set_ylabel('Когнитивные способности')

bins = pd.cut(df['Stress Levels'], bins=[0, 2.5, 4, 5.5, 7, 8.5, 10])
heat = df.pivot_table(values='Cognitive Function', index=bins, columns='Income Level',
                      aggfunc='mean', observed=False)[order]
heat.index = [f'{iv.left:g}–{iv.right:g}' for iv in heat.index]   # читаемые подписи интервалов
sns.heatmap(heat, annot=True, fmt='.1f', cmap='YlGnBu', vmin=50, vmax=75,
            cbar_kws={'label': 'Средняя когн. оценка'}, ax=axes[1])
axes[1].tick_params(axis='y', rotation=0)
axes[1].set_title('Средние когнитивные способности')
axes[1].set_xlabel('Уровень дохода')
axes[1].set_ylabel('Интервал уровня стресса')

fig.suptitle('Доход, когнитивные способности и стресс', fontsize=15)
plt.tight_layout()
plt.show()
````

**Вывод:** линии тренда для всех трёх групп дохода почти плоские (наклон от −0.16 до −0.23 балла на единицу стресса) и близки друг к другу, а на тепловой карте средние когнитивные оценки (≈ 62–65) не меняются ни по стрессу, ни по доходу. Значит, в этих данных между доходом, уровнем стресса и самооценкой когнитивных способностей нет заметной связи (главный фактор для когнитивной самооценки в данных, как видно из задания 5, — возраст).
